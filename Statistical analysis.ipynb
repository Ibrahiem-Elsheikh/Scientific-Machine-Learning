```python
import numpy as np
import pandas as pd
import scipy.stats as stats 
import matplotlib.pyplot as plt
print('all libs imported successfully')
```

```python
Temperature = np.array ([175, 188, 192, 169, 169, 184, 181, 198, 173, 189, 186, 194, 179, 177, 183, 191])

print(f"Data : {Temperature}")
print(f"number of samples: {len(Temperature)}")
```

```python
print(f"Mean: {np.mean(Temperature)}")
print(f"Median: {np.median(Temperature)}")
print(f"Variance: {np.var(Temperature,ddof=1)}")
print(f"sta Deviation {np.std(Temperature,ddof=1)}")
```

```python
x = np.array([1,2,3,4,5,]) 
y = np.array([2,4,6,8,10])

print(f"covarience : {np.cov(x,y)}")
print(f"Perason Correlation: {stats.pearsonr(x,y)}")
```

```python
plt.figure(figsize=(10,5),dpi=100)
plt.scatter(x,y,label="Data points")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Relation between x and y")
plt.grid(True)
plt.legend()
plt.show
```

```python
mu = 0 
Sigma = 1 
Data = np.linspace(-4,4,1000)
pdf = stats.norm.pdf(Data,mu,Sigma)
plt.figure(figsize=(20,10), dpi=100)
plt.plot(Data,pdf, label="Normal Distribution")
plt.title("Normal Distribution")
plt.legend()
plt.grid(True)
plt.show()
```
